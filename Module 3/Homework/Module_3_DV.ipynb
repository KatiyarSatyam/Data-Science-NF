```python
import matplotlib.pyplot as plt
import pandas as pd
```

#Dataset

```python
#Data
Months = ["JAN","FEB","MAR","APR","MAY", "JUN","JUL","AUG","SEP","OCT","NOV","DEC"]
Unit = [45,64,53,66,33,21,52,67,14,43,11,98]
Revenue = [3000,5433,4532,5676,2587,2154,4234,5798,1354,3454,1243,8756]
```

#Line Chart

```python
plt.style.use('ggplot')

#Plot
plt.plot(Months, Revenue, marker = "o", linestyle = "--", linewidth = 1, markerfacecolor = "red", color = "black")

#Label
plt.xlabel("Month",alpha = 0.9, color = "black",fontsize = 14)
plt.ylabel("Revenue(in $)", alpha = 0.9, color = "black", fontsize = 14)
plt.title("Revenue per Month", alpha = 0.9, color = "black", fontsize = 16)

#Ticks Rotation
plt.xticks(rotation=45)

#Range in Y
plt.yticks(range(500,9000,1000))

#Grid
plt.grid(True, alpha = 0.9, linestyle = "--")
plt.show()
```

#Bar Chart

```python
plt.style.use("ggplot")

#Bar chart
plt.bar(Months,Revenue, color = ["purple","indigo","blue", "green","yellow", "orange", "red"])

#label
plt.xlabel("Month",alpha = 0.9, color = "black",fontsize = 14)
plt.ylabel("Revenue(in $)", alpha = 0.9, color = "black", fontsize = 14)
plt.title("Revenue per Month", alpha = 0.9, color = "black", fontsize = 16)

#Ticks
plt.xticks(rotation=45)

#limit
plt.ylim(0,9000)

#Number at bars
for index, value in enumerate(Revenue):
  plt.text(index, value+1, str(value), ha = "center",fontsize = 8, color = "red")

#Grid
plt.grid(axis = 'y', linewidth = 0.9, linestyle = "--", color = 'black')
plt.show()
```

#Histogram

```python
#plot units
plt.figure(figsize=(5,3))
plt.bar(Months,Unit, color = ["purple","indigo","blue"], label = 'Unit')
plt.xlabel("Month",alpha = 0.9, color = "black",fontsize = 14)
plt.ylabel("Units", alpha = 0.9, color = "black", fontsize = 14)
plt.title("Units per Month", alpha = 0.9, color = "black", fontsize = 16)
plt.xticks(rotation=45)
plt.legend()
plt.show()

# Plot Revenue
plt.figure(figsize=(5,3))
plt.bar(Months,Revenue, color='orange', label='Revenue')
plt.xlabel("Months",alpha = 0.9, color = "black", fontsize = 16)
plt.ylabel("Revenue",alpha = 0.9, color = "black", fontsize = 16)
plt.title("Monthly Revenue Histogram",alpha = 0.9, color = "black", fontsize = 16)
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

#Scatter Plot

```python
Months = ["JAN","FEB","MAR","APR","MAY", "JUN","JUL","AUG","SEP","OCT","NOV","DEC"]
Unit = [45,64,53,66,33,21,52,67,14,43,11,98]
Revenue = [3000,5433,4532,5676,2587,2154,4234,5798,1354,3454,1243,8756]
size = [300,500,400,560,250,210,420,570,130,340,120,875]
colors=["red","green","black","cyan", "blue", "purple","indigo","green","yellow", "orange","violet", "pink"]
plt.figure(figsize=(11,4))
#Scatter Plot
plt.scatter(Months, Revenue, s = size, c=colors, alpha = 0.8)
plt.colorbar(label = "Intensity")
#Label
plt.xlabel("Months")
plt.ylabel("Revenue (in $)")
plt.title("Revenue per Month")

plt.show()
```

#Pie Chart

```python
#Data
data = {"Months":["JAN","FEB","MAR","APR","MAY", "JUN","JUL","AUG","SEP","OCT","NOV","DEC"],
"Unit":[45,64,53,66,33,21,52,67,14,43,11,98],
"Revenue": [3000,5433,4532,5676,2587,2154,4234,5798,1354,3454,1243,8756]}

#Sum of Profit
q1 = sum(data["Unit"][0:3])
q2 = sum(data["Unit"][3:6])
q3 = sum(data["Unit"][6:9])
q4 = sum(data["Unit"][9:12])

Quarter = ["Q1","Q2","Q3","Q4"]
Profit = [q1,q2,q3,q4]
Rang = ["purple","indigo","blue", "cyan"]

#Pie Chart
plt.pie(Profit, labels=Quarter, autopct="%1.1f%%", startangle = 53, explode = (0,0,0,0.1), shadow= True, colors = Rang)

#Title
plt.title("Quaterly Profit")

#creating pie chart circle
plt.axis("equal")

#legend
plt.legend(title = "Profit per Quarter", loc = "upper right", fontsize = 8)
plt.show()
```
