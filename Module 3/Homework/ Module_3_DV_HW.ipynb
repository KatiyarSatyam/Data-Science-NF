```python
import matplotlib.pyplot as plt
import pandas as pd
```

#Dataset

```python
#Data
Months = ["JAN","FEB","MAR","APR","MAY", "JUN","JUL","AUG","SEP","OCT","NOV","DEC"]
Unit = [45,64,53,66,33,21,52,67,14,43,11,98]
Revenue = [3000,5433,4532,5676,2587,2154,4234,5798,1354,3454,1243,8756]
```

#Line Chart

```python
plt.style.use('ggplot')

#Plot
plt.plot(Months, Revenue, marker = "o", linestyle = "--", linewidth = 1, markerfacecolor = "red", color = "black")

#Label
plt.xlabel("Months",alpha = 0.9, color = "black",fontsize = 14)
plt.ylabel("Revenue(in $)", alpha = 0.9, color = "black", fontsize = 14)
plt.title("Revenue per Month", alpha = 0.9, color = "black", fontsize = 16)

#Ticks Rotation
plt.xticks(rotation=45)

#Range in Y
plt.yticks(range(500,9000,1000))

#Grid
plt.grid(True, alpha = 0.9, linestyle = "--")
plt.show()
```

#Bar Chart

```python
plt.style.use("ggplot")

#Bar chart
plt.bar(Months,Revenue, color = ["purple","indigo","blue", "green","yellow", "orange", "red"])

#label
plt.xlabel("Months",alpha = 0.9, color = "black",fontsize = 14)
plt.ylabel("Revenue(in $)", alpha = 0.9, color = "black", fontsize = 14)
plt.title("Revenue per Month", alpha = 0.9, color = "black", fontsize = 16)

#Ticks
plt.xticks(rotation=45)

#limit
plt.ylim(0,9000)

#Number at bars
for index, value in enumerate(Revenue):
  plt.text(index, value+1, str(value), ha = "center",fontsize = 8, color = "red")

#Grid
plt.grid(axis = 'y', linewidth = 0.9, linestyle = "--", color = 'black')
plt.show()
```

#Histogram

```python
#plot units
plt.figure(figsize=(5,3))
plt.bar(Months,Unit, color = ["purple","indigo","blue"], label = 'Unit')
plt.xlabel("Months",alpha = 0.9, color = "black",fontsize = 14)
plt.ylabel("Units", alpha = 0.9, color = "black", fontsize = 14)
plt.title("Units per Month", alpha = 0.9, color = "black", fontsize = 16)
plt.xticks(rotation=45)
plt.legend()
plt.show()

# Plot Revenue
plt.figure(figsize=(5,3))
plt.bar(Months,Revenue, color='orange', label='Revenue')
plt.xlabel("Months",alpha = 0.9, color = "black", fontsize = 16)
plt.ylabel("Revenue",alpha = 0.9, color = "black", fontsize = 16)
plt.title("Monthly Revenue Histogram",alpha = 0.9, color = "black", fontsize = 16)
plt.xticks(rotation=45)
plt.legend()
plt.show()
```
