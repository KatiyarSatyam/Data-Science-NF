JOIN IN PANDAS DATAFRAME


```python
import pandas as pd
```

```python
D1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']})
D2= pd.DataFrame( { 'C': [1,2,3,4], 'D': ['a','b','c','d']})
D1.join(D2)
```

D2 IS PRIMARY AND D1 IS SECONDARY

```python
D1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']})
D2= pd.DataFrame( { 'C': [1,2,3,4], 'D': ['a','b','c','d']})
D2.join(D1)
```

INDEX IN JOIN

```python
D1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']}, index= ['a','b', 2 , 'd'])
D2= pd.DataFrame( { 'C': [1,2,3,4], 'D': ['a','b','c','d']})
D1.join(D2)
```

value of C and D in reverse

```python
D1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']}, index= ['a','b', 'c' , 'd'])
D2= pd.DataFrame( { 'C': [1,2,3,4], 'D': ['a','b','c','d']}, index= ['d','c', 'b' , 'a'])
D1.join(D2)
```

INNER

```python
D1= pd.DataFrame( { 'A': [1,2,3,4,5], 'B': [10,20,30,40,50]}, index= ['a','b', 'c','d', 'e'])
D2= pd.DataFrame( { 'C': ['A','B'], 'D': ['E', 'F']}, index= ['c','d'])
D1.join(D2, how= 'inner')
```

OUTER

```python
D1= pd.DataFrame( { 'A': [1,2,3,4,5], 'B': [10,20,30,40,50]}, index= ['a','b', 'c','d', 'e'])
D2= pd.DataFrame( { 'C': ['A','B'], 'D': ['E', 'F']}, index= ['c','d'])
D1.join(D2, how= 'outer')
```

LEFT

```python
D1= pd.DataFrame( { 'A': [1,2,3,4,5], 'B': [10,20,30,40,50]}, index= ['a','b', 'c','d', 'e'])
D2= pd.DataFrame( { 'C': ['A','B'], 'D': ['E', 'F']}, index= ['c','d'])
D1.join(D2, how= 'left')
```

RIGHT

```python
D1= pd.DataFrame( { 'A': [1,2,3,4,5], 'B': [10,20,30,40,50]}, index= ['a','b', 'c','d', 'e'])
D2= pd.DataFrame( { 'C': ['A','B'], 'D': ['E', 'F']}, index= ['c','d'])
D1.join(D2, how= 'right')
```

JOINING SAME NAME TABLES

```python
D1= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['A','B','C','D']})
D2= pd.DataFrame( { 'A': [1,2,3,4], 'B': ['a','b','c','d']})
D1.join(D2, lsuffix= '_left', rsuffix= '_right')
```
